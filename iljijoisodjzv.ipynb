```
    # Cell 1: Basic arithmetic
a = 17
b = 5
print('a + b =', a + b)
print('a - b =', a - b)
print('a * b =', a * b)
print('a / b =', a / b)
print('a // b =', a // b)
print('a % b =', a % b)
```

```
# Cell 2: Powers and roots
import math

for n in range(1, 6):
    print(f'2^{n} = {2 ** n}')

print('sqrt(144) =', math.sqrt(144))
print('5^3 =', math.pow(5, 3))
print('hypotenuse(3, 4) =', math.hypot(3, 4))
```

```
# Cell 3: Trigonometry
import math

for deg in [0, 30, 45, 60, 90]:
    rad = math.radians(deg)
    print(f'{deg:>3}deg -> sin={math.sin(rad):.4f}, cos={math.cos(rad):.4f}')

print('pi =', math.pi)
print('e  =', math.e)
```

```
# Cell 4: Series and sums
n = 100
natural_sum = sum(range(1, n + 1))
print(f'Sum of 1..{n} = {natural_sum}')
print('Gauss formula check:', n * (n + 1) // 2)

squares_sum = sum(i ** 2 for i in range(1, 11))
print('Sum of squares 1..10 =', squares_sum)

geometric = sum(1 / 2 ** k for k in range(20))
print('Geometric series (~2) =', geometric)
```

```
# Cell 5: Factorials, combinatorics and Fibonacci
import math

for k in range(1, 8):
    print(f'{k}! = {math.factorial(k)}')

print('C(10, 3) =', math.comb(10, 3))
print('P(10, 3) =', math.perm(10, 3))

def fib(m):
    a, b = 0, 1
    seq = []
    for _ in range(m):
        seq.append(a)
        a, b = b, a + b
    return seq

print('First 10 Fibonacci:', fib(10))
```
