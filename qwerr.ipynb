```python
import numpy as np
import tensorflow as tf
import torch
from datasets import Dataset

np_arr = np.array([[1, 2, 3], [4, 5, 6]])
np_arr
```

```python
np_rec = np.rec.array([(1, 2.5, 'a'), (3, 4.5, 'b')], dtype=[('x', 'i4'), ('y', 'f4'), ('z', 'U1')])
np_rec
```

```python
tf_tensor = tf.constant([[1, 2], [3, 4]])
tf_tensor
```

```python
tf_var = tf.Variable([[1.0, 2.0], [3.0, 4.0]])
tf_var
```

```python
tf_sparse = tf.SparseTensor(indices=[[0, 0], [1, 2]], values=[1, 2], dense_shape=[3, 4])
tf_sparse
```

```python
torch_tensor = torch.tensor([[1, 2], [3, 4]])
torch_tensor
```

```python
hf_data = {"name": ["Alice", "Bob", "Charlie"], "age": [25, 30, 35]}
hf_dataset = Dataset.from_dict(hf_data)
hf_dataset
```
