```
# 5. Linear Algebra — matrix operations with numpy
import numpy as np

A = np.array([[3, 1, 2], [2, 4, 1], [1, 2, 5]])
B = np.array([[1, 0, 1], [0, 2, 1], [1, 1, 0]])

print('A =\n', A)
print('B =\n', B)
print('A @ B =\n', A @ B)
print('det(A) =', round(np.linalg.det(A), 6))
print('A^-1 =\n', np.round(np.linalg.inv(A), 4))
vals, vecs = np.linalg.eig(A)
print('Eigenvalues of A:', np.round(vals, 4))
```

```
# 4. Statistics with numpy
import numpy as np

np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)

print(f'Mean:     {np.mean(data):.4f}')
print(f'Median:   {np.median(data):.4f}')
print(f'Std dev:  {np.std(data):.4f}')
print(f'Variance: {np.var(data):.4f}')
print(f'Min/Max:  {data.min():.4f} / {data.max():.4f}')
print(f'25th pct: {np.percentile(data, 25):.4f}')
print(f'75th pct: {np.percentile(data, 75):.4f}')
```

```
# 3. Calculus — derivatives & integrals with sympy
from sympy import symbols, diff, integrate, sin, cos, exp, pi, oo

x = symbols('x')

f = x**3 * sin(x)
print('f(x)  =', f)
print("f'(x) =", diff(f, x))
print("f''(x)=", diff(f, x, 2))

print('∫ x^3 dx =', integrate(x**3, x))
print('∫₀^π sin(x) dx =', integrate(sin(x), (x, 0, pi)))
print('∫₀^∞ e^-x dx =', integrate(exp(-x), (x, 0, oo)))
```

```
# 2. Algebra — solving equations with sympy
from sympy import symbols, solve, expand, factor

x, y = symbols('x y')

eq1 = x**2 - 5*x + 6
print('Roots of x^2 - 5x + 6:', solve(eq1, x))

system = [2*x + y - 5, x - 3*y + 1]
print('System solution:', solve(system, [x, y]))

expr = (x + 2)**3
print('Expand (x+2)^3:', expand(expr))
print('Factor x^3+6x^2+12x+8:', factor(expr))
```
