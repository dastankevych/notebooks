Задача: экспортировать JSON (small.json) в CSV и прочитать CSV как DataFrame.

```python
import pandas as pd
import json
import os

json_path = "small.json"
csv_path = "small_export.csv"

with open(json_path, "r") as f:
    data = json.load(f)

df_json = pd.json_normalize(data, sep=".")
df_json.to_csv(csv_path)
df_json
```
